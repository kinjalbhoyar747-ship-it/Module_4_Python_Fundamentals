## 1. Why loops? (8 min)

If we need to inspect ten temperature readings, copying an `if` statement ten times is slow and error-prone. A loop says: *for each item, carry out these steps.*

We will use `for` loops first because they are the safest starting point when the number of repetitions is known.

```python
# for i in range (5) gives the value 0, 1, 2, 3, 4
# for i in range (1, 6) here 1 is the start  and 6 is the end value, so it gives the value 1, 2, 3, 4, 5
# for i in range (1, 6, 2) here 1 is the start, 6 is the end value and 2 is the step value, so it gives the value 1, 3, 5
# range is the function.
# even we want to print even numbers from 0 to 10 we can use range function like this range(0, 11, 2) here 0 is the start value, 11 is the end value and 2 is the step value, so it gives the value 0, 2, 4, 6, 8, 10
# agar (11,2) ttype kiya toh 11 se start hoga aur 2 se step hoga, toh 11, 9, 7, 5, 3, 1 print hoga
# agar odd numbers print karne ho toh range(1, 11, 2) use karenge, toh 1, 3, 5, 7, 9 print hoga
# for reverse:- (10, 2, -2) 
for minute in range(1, 6):
    print(f'Reading at minute {minute}')
```

```python
# for i in range(1, 6):
# m= 0 
# m = m + 250 
# print(i,m)
# 5 ghante ke liye flow rate measure karna hai, har ghante 250 ml flow rate badh rha hai, toh total flow rate kitna hoga?
m = 0
for i in range(1, 6):
    m = m + 250
    print(i, m)
```

`range(1, 6)` creates 1, 2, 3, 4, 5: the final number is excluded.

## 2. Repeated calculations (15 min)

Indent all instructions that should repeat. Here, the same conversion calculation is applied at several times.

```python
for time_min in range(0, 31, 5):
    conversion_percent = 100 * (1 - 0.9**time_min)
    print(f'{time_min:2d} min → conversion = {conversion_percent:6.2f}%')
```

## 3. `while` loops: repeat until a condition changes (12 min)

A `while` loop repeats as long as its condition remains true. Make sure something inside the loop changes; otherwise the loop can run forever. For early learning, prefer `for` when possible.

```python
tank_level_percent = 100

while tank_level_percent > 20:
    print(f'Current level: {tank_level_percent}%')
    tank_level_percent = tank_level_percent - 20

print('Low-level threshold reached.')
```

```python
# water level = 100 jo kam ho rha hai with time. 20 ke neeche water level nhi jaana chahiye. toh while loop use karenge.
water_level = 100
while water_level > 20:
    print(f'OKAY.')
    water_level = water_level - 20
```

```python
# to print 0 to 10 even numbers 
n = 0
while n <= 10:
    print (n)
    n = n + 2
```

## 4. Combining loops and decisions (15 min)

A useful pattern is: loop through readings, then use a conditional to label each one. The code below is a simple alarm scan—not a real safety system.

```python
temperatures_c = [72, 84, 96, 88, 101]

for temperature_c in temperatures_c:
    if temperature_c > 100:
        label = 'ALARM'
    elif temperature_c > 85:
        label = 'CAUTION'
    else:
        label = 'normal'
    print(f'{temperature_c} °C → {label}')
```

```python
# array is a collection of values, in python it is called list. Also called a container. It can store multiple values in a single variable.
# example of list is temperatures_c = [72, 84, 96, 88, 101] here 5 values are stored in a list. First element ka index 0 hota hai, second element ka index 1 hota hai, third element ka index 2 hota hai, fourth element ka index 3 hota hai, fifth element ka index 4 hota hai.
# print (temperatures_c[0]) likha toh 72 print hoga, print (temperatures_c[1]) likha toh 84 print hoga, print (temperatures_c[2]) likha toh 96 print hoga, print (temperatures_c[3]) likha toh 88 print hoga, print (temperatures_c[4]) likha toh 101 print hoga.
# peeche se index -1 hota hai. print (temperatures_c[-1]) likha toh 101 print hoga, print (temperatures_c[-2]) likha toh 88 print hoga, print (temperatures_c[-3]) likha toh 96 print hoga, print (temperatures_c[-4]) likha toh 84 print hoga, print (temperatures_c[-5]) likha toh 72 print hoga.
#mutable list hai, toh hum list ke elements ko change kar sakte hai. Example: temperatures_c[0] = 75 likha toh 72 ki jagah 75 aa jayega. print (temperatures_c) likha toh [75, 84, 96, 88, 101] print hoga.
# append function use karke list me new element add kar sakte hai. Example: temperatures_c.append(110) likha toh 110 list me add ho jayega. print (temperatures_c) likha toh [75, 84, 96, 88, 101, 110] print hoga.
# pop function use karke list me se last element remove kar sakte hai. Example: temperatures_c.pop() likha toh 110 list me se remove ho jayega. print (temperatures_c) likha toh [75, 84, 96, 88, 101] print hoga.


temperatures_c = [20, 50, 70, 30, 60, 80]
for temperature_c in temperatures_c:
    if temperature_c >= 65:
        label = 'Alarm'
    elif temperature_c > 50:
        label = 'Near Ok'
    else:
        label = 'Normal'
    print(f'{temperature_c} °C → {label}')
```
